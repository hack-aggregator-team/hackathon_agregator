```python
import json
import requests
import re

with open('hacks.txt', 'r', encoding='utf-8') as f:
    hacks = json.load(f)
hack_links = [f"https://www.xn--80aa3anexr8c.xn--p1acf/tpost/{post['uid']}" for post in hacks['posts']]

api_url = "https://feeds.tildaapi.com/api/getfeed/?feeduid=617755803461&recid=488755787&c=1778524652766&size=8&slice=1&sort%5Bdate%5D=desc&filters%5Bdate%5D%5B%3E%5D=now&getparts=true"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept": "application/json",
}

response = requests.get(api_url, headers=headers)

if response.status_code == 200:
    data = response.json()
    print("Успех! Получено данных:", len(data))
    with open("mock_hackathons.json", "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=4)
    print("Файл mock_hackathons.json успешно сгенерирован!")
else:
    print(f"Что-то пошло не так. Код ошибки: {response.status_code}")

import requests
from bs4 import BeautifulSoup


headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
    "Accept-Language": "ru-RU,ru;q=0.9,en-US;q=0.8,en;q=0.7",
}
soups = []
for url in hack_links:
    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        soups.append(soup)
        print(f"Парсинг {url.split('/')[-1]} успешен")
    else:
        print(f"Ошибка доступа: {response.status_code}")
        break

raw_data = [soup.find('div', class_='t-redactor__tte-view').find_all('div', class_='t-redactor__text') for soup in soups]

parsed_hackathons = []
for item in raw_data:
    hack_data = {
        "title": "",
        "prize": "Не указан",
        "dates": "Не указаны",
        "format": "Не указан",
        "deadline": "Не указана",
        "description": ""
    }

    if item:
        hack_data["title"] = item[0].get_text(strip=True)

    for block in item[1:]:
        soup = BeautifulSoup(str(block), 'html.parser')

        for li in soup.find_all('li'):
            li.insert_before('• ')

        text = soup.get_text(separator=" ", strip=True)
        text_lower = text.lower()

        if "хочешь узнавать" in text_lower or "telegram-бот" in text_lower:
            continue

        if "призовой фонд" in text_lower:
            val = re.split(r'(?i)призовой фонд(?:[\s:—-]+)', text)[-1]
            hack_data["prize"] = val.strip()

        elif "дат" in text_lower and "проведения" in text_lower or "когда" in text_lower:
            val = re.split(r'(?i)(?:даты проведения|дата проведения|когда\?)(?:[\s:—-]+)', text)[-1]
            hack_data["dates"] = val.strip()

        elif "формат" in text_lower:
            val = re.split(r'(?i)формат(?:[\s:—-]+)', text)[-1]
            hack_data["format"] = val.strip()

        elif "регистрация до" in text_lower:
            match = re.search(r'(?i)регистрация до\s*(.*?)(?::|$)', text)
            if match:
                hack_data["deadline"] = match.group(1).strip()

        else:
            desc_chunk = soup.get_text(separator="\n", strip=True)
            hack_data["description"] += desc_chunk + "\n\n"

    hack_data["description"] = hack_data["description"].strip().replace("\n•\n", "\n• ")
    parsed_hackathons.append(hack_data)

with open("mock_hackathons.json", "w", encoding="utf-8") as f:
    json.dump(parsed_hackathons, f, ensure_ascii=False, indent=4)
```
